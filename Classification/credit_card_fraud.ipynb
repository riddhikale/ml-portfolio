## Conclusion

This dataset had a much bigger imbalance problem than anything before it — 
only 0.97% of transactions were fraud, compared to Bank Churn's 20%. That 
means a model which just guessed "not fraud" every single time would already 
be 99.03% "accurate" without actually learning anything. So accuracy alone 
means almost nothing here.

Logistic Regression still did really well as a first try (Recall 0.8265, 
Precision 0.9759). This is probably because the V1-V28 columns are already 
PCA-transformed — PCA's whole job is to find the directions where data 
differs the most, and since fraud vs non-fraud is a big difference in this 
data, PCA likely picked up on that automatically, even before I trained 
anything.

I lowered the threshold from 0.5 to 0.3, which improved recall to 0.8673 — 
catching more real fraud cases, at a small cost to precision (0.9341). Since 
missing fraud means real money lost, this trade was worth it. But the 
improvement was smaller than in my other projects, which tells me the model 
was already doing a pretty good job separating fraud from real transactions 
even before I touched the threshold.

Unlike Bank Churn, I decided not to try a Decision Tree here. With recall at 
86.73% and precision still above 93%, Logistic Regression is already doing 
well, and it's simpler and easier to explain than a more complex model. Since 
the gains from tuning were already small, adding a more complex model didn't 
feel worth it for this project. Sometimes the simpler model that already 
works well is the right choice, not the most powerful one available.
