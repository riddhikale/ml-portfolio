```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/creditcard_sample.csv')
df.head()
```

```python
X = df.drop(columns=['Class'])
y = df['Class']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.decomposition import PCA

pca = PCA()
pca.fit(X_scaled)
```

```python
print("Cumulative variance:", pca.explained_variance_ratio_.cumsum())
```

```python
plt.plot(range(1, len(pca.explained_variance_ratio_)+1), pca.explained_variance_ratio_.cumsum(), marker='o')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Variance Explained')
plt.title('PCA Scree Plot - Credit Card Fraud')
plt.axhline(y=0.9, color='r', linestyle='--', label='90% variance')
plt.legend()
plt.show()
```

```python
pca_10 = PCA(n_components=10)
X_pca_10 = pca_10.fit_transform(X_scaled)
print(f"Variance retained with 10 components: {pca_10.explained_variance_ratio_.sum():.4f}")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_pca_10, y, test_size=0.2, random_state=42, stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_train)
y_pred = log_reg.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall: {recall_score(y_test, y_pred):.4f}")
print(f"F1: {f1_score(y_test, y_pred):.4f}")
```

```python
fraud_idx = y_test[y_test == 1].index[0]
legit_idx = y_test[y_test == 0].index[0]

fraud_sample = X_test[y_test.index.get_loc(fraud_idx)].reshape(1, -1)
legit_sample = X_test[y_test.index.get_loc(legit_idx)].reshape(1, -1)
```

```python
for name, sample in [('Actual Fraud Case', fraud_sample), ('Actual Legit Case', legit_sample)]:
    pred = log_reg.predict(sample)
    proba = log_reg.predict_proba(sample)
    print(f"{name}")
    print(f"Predicted: {'Fraud' if pred[0] == 1 else 'Not Fraud'}")
    print(f"Probability: {proba}")
    print()
```
