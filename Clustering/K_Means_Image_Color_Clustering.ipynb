```python
import numpy as np
from PIL import Image
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
img = Image.open('color_clustering.jpeg')
img_array = np.array(img)
```

```python
print("Image shape:", img_array.shape)
```

```python
pixels = img_array.reshape(-1, 3)
print("Pixels shape:", pixels.shape)
```

```python
K = 5
kmeans = KMeans(n_clusters=K, random_state=42, n_init=10)
kmeans.fit(pixels)
```

```python
compressed_pixels = kmeans.cluster_centers_[kmeans.labels_].astype('uint8')
compressed_img = compressed_pixels.reshape(img_array.shape)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(img_array)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(compressed_img)
axes[1].set_title(f'Compressed to {K} Colors (K-Means)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
